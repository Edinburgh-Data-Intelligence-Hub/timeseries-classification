```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
fontsize = 14
fontsize_minor = fontsize-2
```

```python
loss_df = pd.read_csv(f'./results/ts2vec/loss_ts2vec.csv') # Generated by src.modeling.train_ts2vec.py script

fig, ax = plt.subplots(1, figsize=(5,5))

b = sns.lineplot(data=loss_df, x='Epoch', y='Loss', label='train_loss', ax=ax)

b.set(ylim=(0.2, 0.5))

b.set_xlabel(b.get_xlabel(), fontsize=fontsize)
b.set_ylabel(b.get_ylabel(), fontsize=fontsize)

b.set_xticklabels(b.get_xticklabels(), fontsize=fontsize_minor)
b.set_yticklabels(b.get_yticklabels(), fontsize=fontsize_minor)

b.set_title('TS2Vec train loss', fontsize=fontsize)

b.legend_.remove()
sns.despine(top=True, right=True)
```
